# FIT-IC (Flexible Inversion Tool for Inventory Compilers) GUI

## Introduction ##
The FIT-IC tool is designed to address multiple purposes
1. **forward simulation**: assessing the degree of consistency of atmospheric observations with a given (prior)
source/sink scenario
2. **inversion**: inferring a (posterior) flux scenario that is (in a statistical sense) consistent with the prior
scenario and the atmospheric observations
3. **uncertainty propagation**: provision of uncertainty ranges for the inferred posterior flux scenario.
4. **sensitivity analysis**: exploring the effect of differing prior source/sink scenarios on (forward) simulated concentrations

The FIT-IC is implemented in a flexible and scalable manner and will allow access through a web interface as well as on a local computer.
In it's current implementation the focus is on Methane in 2021.

In the following we will shortly describe the usage when the tool is being accessed via a web interface,
a suitable environment is installed on the [ICOS Jupyter Lab](https://jupyter.icos-cp.eu).

## Fit statistics

The fit statistics allows an overview of comparison metrics against observed concentrations as well as simulated (or observed) concentrations across all stations.  
On start-up the simulation with the standard configuration (designated as 'default') is shown with the root mean squared error as selected metric.  
Other experiments are available from the pop-down list and further metrics can be selected by the tabs on the top, as indicated by the red arrows below:

![fit-ic statistics, start](screenshots_fitic-gui/fitic_fit-statistics_sceen2.png)

### visualisation on global map
This statistics overview is split vertically into two panels, where the upper panel visualises the data as scatter plot on a global map and only one experiment can be selected at a time.  
The plots in the FIT-IC GUI generally provide further interactive features. If you drag your mouse or pointer over a selected station on the map, further information is being shown.  
The toolbar on the right-hand side of the plot (see the red arrow) offers different modes of zooming, cropping, and downloading the graphics.

![fit-ic statistics map](screenshots_fitic-gui/fitic_fit-statistics_sceen-upper-panel.png)

### visulisation as dot graph
In contrast, the lower panel shows the selected metric (or simulation) as a dot plot with the stations on the x-axis.  
This representation also allows to compare multiple experiments at the same time.  
The screenshot below shows the bias between simulated and observed concentrations for the standard experiment and a simulation that was run without emissions over Northern America.  
Further experiments could be selected from the list at the right-hand side of the plot (see again the red arrow). 
The stations (on the x-axis) are sorted in increasing order according to the primal experiment selected from the pop-down menu on the top of the statistics viewer window.

## Modelled time-series

We now come to another way of comparing the simulated concentrations with observations, which is accessed by clicking the "modelled time-series" tab (always located the upper left part of the panel for the precomputed experiments):

!['fit-ic, click modelled time-series](screenshots_fitic-gui/fitic_precomp-desc1_cropped.png)

### time-series initial screen
This panel is intended to provide more detailed comparisons between simulation and observations at a selected station.  
Since no experiment is selected on start-up only the observed concentrations for an initially selected station are shown at beginning:

![fit-ic, station explorer start-up](screenshots_fitic-gui/fitic_station-explorer_start.png)

Essentially this window is vertically split into three panels
1. selection of station and experiment(s)
2. visualisation of (observed) concentration time-series
3. short information about currently selected station including a map highlighting it's location

Other stations can be selected from the drop-down menu, but only one station can be selected at a time.  
Contrary several experiments can be selected at the same time. Typically this is achieved by pressing the \<ctrl\> key during selection with the mouse or pointer.

### time-series after selecting experiments
For the screenshot below two experiments have been selected, again the experiment with the standard emission setup and an experiment without any emissions in the innermost simulation's zoom domain which roughly covers Germany, Netherlands, and Switzerland.

![fit-ic, station explorer start-up](screenshots_fitic-gui/fitic_station-explorer_selected-experiments.png)

- In the upper panel, next to the configuration widgets, a summary table is shown.  
  This lists, per experiment, all derived metrics (root-mean squared error, bias, and Pearson correlation coefficent) derived from the complete time-series.
- The middle panel now contains two plots:
  - the first just shows the time-serieses of observed and simulated concentrations
  - the second visualises the weekly-averaged bias (difference between simulated and observed concentration)
- In the lower panel (on the right to information about the station) additionally a histogram of the hourly biases is shown
  and provides an indication on how these biases are distributed.
